```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#********************************************  PANDAS  ********************************************************
data = pd.read_csv("girls_education.csv")
print(data)
print(data.head(10))
print(data.shape)
print(data.columns)
print(data.dtypes)
print(data.describe())
#Display only the Student_ID, Location, Household_Income, and Attendance_Percent columns.
print(data[["Student_ID", "Location", "Household_Income","Attendance_Percent"]])
#Find all students whose Attendance_Percent is greater than 90%.
print(data[data["Attendance_Percent"] > 90])
#Find students whose: Parents_Education is Graduate and Household_Income is greater than 60,000.
print(data[(data["Parents_Education"] == "Graduate") & (data["Household_Income"] > 60000)])
#Sort the students by Attendance_Percent from highest to lowest and display the top 5 students.
print(data.sort_values("Attendance_Percent",ascending = False).head(5))
#Find students who live in a Rural area, have No school access.
print(data[(data["Location"] == "Rural") & (data["School_Access"] == "No")])

#****************************************   NUMPY   **********************************************************
#Convert the Attendance_Percent column into a NumPy array
Attendance_array = data["Attendance_Percent"].to_numpy()
print(Attendance_array)
#calculate
print(Attendance_array.mean())       #mean
print(np.median(Attendance_array))   #median
print(Attendance_array.std())        #standard deviation
print(Attendance_array.min())        #minimum value
print(Attendance_array.max())        #maximum value
#Using NumPy, determine how many students have attendance above the overall average attendance.
average = Attendance_array.mean()
print(Attendance_array.mean())    
count = np.sum(Attendance_array > average)
print(count)
#Convert Household_Income into a NumPy array.
Household_Income_array = data["Household_Income"].to_numpy()
print(Household_Income_array)
#calculate:
#Total household income
print(np.sum(Household_Income_array))
#Average household income
print(Household_Income_array.mean())
#Maximum income
print(Household_Income_array.max())


#*******************************************  MATPLOTLIB  *****************************************************
#Create a line plot showing Distance_km vs Attendance_Percent
Distance = data["Distance_km"].to_numpy()
print("Distance : ",Distance)
Attendance = data["Attendance_Percent"].to_numpy()
print("Attendance : ",Attendance)
plt.plot(Distance , Attendance, marker = "o", color = "red")
plt.title("Factors affecting girl's education")
plt.xlabel("Distance")
plt.ylabel("Attendance(%)")
plt.grid(True)
plt.show()

#Create a bar chart showing the number of students in each Location.
location_count = data["Location"].value_counts()
print(location_count)
plt.bar(location_count.index, location_count.values ,color = ["yellow","green"])
plt.title("Number of Students by Location")
plt.xlabel("Location")
plt.ylabel("Number of Students")
plt.show()

#Create a scatter plot showing the relationship between Household_Income and Attendance_Percent.
Household_Income = data["Household_Income"].to_numpy()
print("Household_Income : ",Household_Income)
Attendance = data["Attendance_Percent"].to_numpy()
print("Attendance : ",Attendance)
plt.scatter(Household_Income , Attendance, color = "red")
plt.title("Household_Income vs Attendance_Percent")
plt.xlabel("Household_Income")
plt.ylabel("Attendance(%)")
plt.grid(True)
plt.show()

#Create a histogram showing the distribution of Attendance_Percent.
Attendance = data["Attendance_Percent"].to_numpy()
print("Attendance : ",Attendance)
plt.hist(Attendance, bins=4 , color="yellow", edgecolor="black" )
plt.title("Distribution of Attendance_Percent")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")
plt.show()


#------------------------------------------------------------------------------------------------------------------
#show full dataset in pdf
print(data.to_string())


```
